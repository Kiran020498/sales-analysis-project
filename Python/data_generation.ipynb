```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
date_range = pd.date_range(
    start = "2022-01-01",
    end = "2024-12-31"
) #date generation
```

```python
calender = pd.DataFrame({
    "date": date_range,
    "year" : date_range.year,
    "month" : date_range.month,
    "quarter" : date_range.to_period("Q").astype(str)
}) #extracting components from datetime
```

```python
calender.to_csv(r"C:\Users\Kiran Kumar\sales-analysis-project\Data\Data-Raw\calendar.csv", index=False)
```
